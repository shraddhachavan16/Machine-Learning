```python
import pandas as pd

df = pd.read_csv("titanic.csv")
print(df.info())
```

```python
df_before = df.copy()

# Drop useless columns
df_before = df_before.drop(['Name','Ticket','Cabin'], axis=1)

# Convert categorical to numeric
df_before['Sex'] = df_before['Sex'].map({'male':0, 'female':1})
df_before['Embarked'] = df_before['Embarked'].map({'S':0, 'C':1, 'Q':2})

# Drop missing rows (since we are NOT cleaning)
df_before = df_before.dropna()

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X = df_before.drop('Survived', axis=1)
y = df_before['Survived']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

acc_before = accuracy_score(y_test, y_pred)
print("Accuracy BEFORE cleaning:", acc_before)
```

```python
df_clean = df.copy()

df_clean['Age'].fillna(df_clean['Age'].median(), inplace=True)
df_clean['Fare'].fillna(df_clean['Fare'].mean(), inplace=True)
df_clean['Embarked'].fillna(df_clean['Embarked'].mode()[0], inplace=True)
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
print(df_clean.duplicated().sum())
```

```python
Q1 = df_clean['Fare'].quantile(0.25)
Q3 = df_clean['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_clean = df_clean[(df_clean['Fare'] >= lower) & (df_clean['Fare'] <= upper)]
```

```python
import numpy as np

df_clean['Fare'] = np.log1p(df_clean['Fare'])
```

```python
df_after = df_clean.drop(['Name','Ticket','Cabin'], axis=1)

df_after['Sex'] = df_after['Sex'].map({'male':0, 'female':1})
df_after['Embarked'] = df_after['Embarked'].map({'S':0, 'C':1, 'Q':2})

X = df_after.drop('Survived', axis=1)
y = df_after['Survived']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

acc_after = accuracy_score(y_test, y_pred)
print("Accuracy AFTER cleaning:", acc_after)
```

```python
print("Before Cleaning:", acc_before)
print("After Cleaning:", acc_after)
```
